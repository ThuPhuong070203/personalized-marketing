```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Thêm root dir
ROOT_DIR = Path("..").resolve().parent
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))
import config

print("Project Root:", config.PROJECT_ROOT)
```

```python
# Đọc RFM train
rfm = pd.read_parquet(config.RFM_TRAIN_PATH)
print(f"Tổng số khách hàng: {len(rfm):,}")
rfm.describe().T
```

```python
# Vẽ phân bố log1p RFM
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].hist(np.log1p(rfm['recency']), bins=30, color='skyblue', edgecolor='black')
axes[0].set_title('Log1p(Recency)')

axes[1].hist(np.log1p(rfm['frequency']), bins=30, color='salmon', edgecolor='black')
axes[1].set_title('Log1p(Frequency)')

axes[2].hist(np.log1p(rfm['monetary']), bins=30, color='lightgreen', edgecolor='black')
axes[2].set_title('Log1p(Monetary)')
plt.tight_layout()
plt.show()
```
